```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle
import matplotlib.gridspec as gridspec
```

## num1:ath

```python
species = 'ath'

# ===== 读取 rRNA 区域信息 =====
rRNA_info = pd.read_csv(f'{species}/rRNA.info',sep='\t')

# 去掉 5S 区域
rRNA_info = rRNA_info[rRNA_info['type'] != 'rRNA_5S']

# ===== 颜色 =====
color_dict = {
    'rRNA_18S': '#377eb8',
    'ITS1': '#ff7f00',
    'rRNA_5_8S': '#984ea3',
    'ITS2': '#e41a1c',
    'rRNA_25S': '#a65628'
}

# ===== 读取数据 =====
normalized_df = pd.read_csv(f'{species}.csv', index_col=0)

# 去掉 heatmap 中 5S 对应的列
valid_positions = set()
for _, row in rRNA_info.iterrows():
    valid_positions.update(range(row['start_pos'], row['end_pos'] + 1))

# 保留这些位置
normalized_df = normalized_df.loc[:, normalized_df.columns.astype(int).isin(valid_positions)]

# ===== 画图 =====
fig = plt.figure(figsize=(12, 8))

gs = gridspec.GridSpec(
    2, 1,
    height_ratios=[1.2, 20],
    hspace=0.02
)

# ===== heatmap =====
ax = fig.add_subplot(gs[1])

sns.heatmap(
    normalized_df,
    ax=ax,
    cmap="Reds",
    cbar_kws={
        'label': 'Normalized Depth',
        'ticks': [0, 250000, 500000, 750000, 1000000]
    },
    xticklabels=200,
    yticklabels=50,
    vmin=0,
    vmax=1000000
)

ax.set_xlabel("rRNA Positions")
ax.set_ylabel("Samples")
ax.set_title("Samples reads Enrichment in ATH rRNA (no 5S)")

xmin, xmax = ax.get_xlim()

# ===== annotation bar =====
ax_anno = fig.add_subplot(gs[0], sharex=ax)

for _, row in rRNA_info.iterrows():

    start = row['start_pos'] - 0.5
    end = row['end_pos'] - 0.5
    width = end - start

    region_type = row['type']
    color = color_dict.get(region_type, "lightgrey")

    ax_anno.add_patch(
        Rectangle(
            (start, 0),
            width,
            1,
            facecolor=color,
            edgecolor="white",
            linewidth=1
        )
    )

    label = region_type.replace("rRNA_", "")

    if width > 50:
        ax_anno.text(
            start + width / 2,
            0.5,
            label,
            ha='center',
            va='center',
            fontsize=10
        )

ax_anno.set_xlim(xmin, xmax)
ax_anno.set_ylim(0, 1)
ax_anno.axis("off")

fig.subplots_adjust(hspace=0.02)

plt.savefig(
    "plot_rRNA_depth.ath.no5S.pdf",
    format="pdf",
    bbox_inches="tight"
)

plt.show()
```
